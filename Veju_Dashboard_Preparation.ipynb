**Objective:** Validate KPI metrics, compare Bengaluru vs. Chennai climate patterns, and prepare summary aggregations for the Power BI Dashboard.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("AtmoSync_Bengaluru_Chennai_Weather_2025.csv")
print(f"Total Rows: {len(df)} | Columns: {len(df.columns)}")
print("Cities:", df["City"].unique().tolist())
```

```python
# Verify Top KPI Card Values for Power BI
print("--- OVERALL KPI SUMMARY ---")
print("Average Temperature (°C):", round(df["Temperature"].mean(), 1))
print("Average Humidity (%):", round(df["Humidity"].mean(), 1))
print("Average Wind Speed (km/h):", round(df["Wind_Speed"].mean(), 1))
print("Total Rainfall (mm):", round(df["Rainfall"].sum(), 1))
```

```python
# Bengaluru vs. Chennai Comparison Table
print("--- BENGALURU VS CHENNAI COMPARISON ---")
comparison = df.groupby("City").agg({
    "Temperature": "mean",
    "Humidity": "mean",
    "Wind_Speed": "mean",
    "Rainfall": "sum"
}).round(1)
print(comparison)
```

```python
# Seasonal Monsoon Shift Verification (SW Monsoon vs NE Monsoon)
print("--- SEASONAL RAINFALL BREAKDOWN (mm) ---")
seasonal_rain = df.pivot_table(index="City", columns="Season", values="Rainfall", aggfunc="sum").round(1)
print(seasonal_rain)
```
